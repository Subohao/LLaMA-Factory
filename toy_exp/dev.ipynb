```python
import json

with open('toy_audio_toolcall.json', 'r') as f:
    data = json.load(f)

print(data[0])
```

```python
import re
matches = re.findall(r"<audio>(.*?)</audio>", data[0]['conversations'][0]['value'])
print(matches)
```

```python
import os
import re
new_format_data = []

for item in data:
    new_item = item.copy()
    audio_path = re.findall(r"<audio>(.*?)</audio>", new_item['conversations'][0]['value'])[0]
    new_item['conversations'][0]['value'] = new_item['conversations'][0]['value'].replace(f"<audio>{audio_path}</audio>", "<audio>")
    if audio_path.startswith('must_c'):
        new_item['audios'] = [os.path.join('/ocean/projects/cis210027p/bsu5/LLaMA-Factory/toy_exp', audio_path)]
    else:
        new_item['audios'] = [audio_path]
    new_format_data.append(new_item)

with open('toy_audio_toolcall_new_format.json', 'w') as f:
    json.dump(new_format_data, f, indent=4)
```

```python
audios = []
for dia in new_format_data:
    audios.extend(dia['audios'])

print(len(audios))
print(len(set(audios)))
```

```python
import os
import shutil

# Destination directory
dst_dir = 'librimix_audios'
os.makedirs(dst_dir, exist_ok=True)

for audio_path in audios[:300]:
    if os.path.isfile(audio_path):
        shutil.copy(audio_path, dst_dir)
        # print(f"Copied {audio_path} to {dst_dir}")
    else:
        print(f"File not found: {audio_path}")
```
